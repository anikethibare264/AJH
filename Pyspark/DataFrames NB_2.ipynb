```python
from pyspark.sql import functions as f
```

```python
df=(
    spark.read.format("csv")
.option("header","true")
.option("inferSchema","true")
.load("/Volumes/workspace/default/csv_data/diamonds.csv"))
display(df)
```

```python
df.columns
```

```python

# aggregation functions(Single)
new_df=(
    df
    .groupby('cut')
    .count()
)
display(new_df)
```

```python

new_df=(
    df.groupBy('clarity').count()
)
display(new_df)
```

```python
# aggregation functions(Multiple)
new_df=(
    df
    .groupBy('cut')
    .agg(
        f.count('*'). alias('T-Records'),
        f.max(f.col("price")). alias('max-price')
    )
)
display(new_df)
```

```python
df=(
    spark.read.format("csv")
.option("header","true")
.option("inferSchema","true")
.load("/Volumes/workspace/default/csv_data/Students Social Media Addiction.csv"))
display(df)
```

```python
new_df=(
    df
    .groupBy('Academic_Level')
    .agg(
        f.count('*'). alias('T-Records')
        )
)
display(new_df)
```

```python
from pyspark.sql import functions as F
new_df=(
    df
    .groupBy("Country")
    .agg(
       F.collect_list('Student_ID').alias("STU_List")    )   #U can use collect_set for non_ duplicate values
)
display(new_df)
```
