```python
# 1. Creating a data frame with no strict schema
my_data=[
    (21,'Santosh',85),
    (22,'Ravi',80),
    (23,'Rajesh',75),
    (24,'Rakesh',70),
    (25,'Raj',65)
]
collumns=['stu_id','name','marks']
stu_df=spark.createDataFrame(data=my_data,schema=collumns)
display(stu_df)
```

```python
# 2. creating the data frame with schema
import pyspark.sql.types as t
from pyspark.sql.types import StructType,StructField,StringType,IntegerType
emp_data=[
    (101,"Ravi","IT",8000),
    (102,"Rajesh","IT",9000),
    (103,"Rakesh","HR",10000),
    (104,"Raj","HR",12000),
    (105,"Santosh","IT",15000)
]

emp_schema=StructType([
    StructField("emp_id",IntegerType(),False),
    StructField("name",StringType(),True),
    StructField("dept",StringType(),True),
    StructField("salary",IntegerType(),True)
])

emp_df=spark.createDataFrame(data=emp_data,schema=emp_schema)
display(emp_df)

```

```python
# Handlin the null values for the we need to create one bcoz we dont have the null values

new_df=(
    df.
    withColumn(
        "Is_Active",
        f.when(
            f.col("Student_ID")%2!=0,
            f.lit(None)
        ).otherwise(1)
        )
)

new_df.select("Student_ID","Age","Is_Active").display()
```

```python
# it will delete all the null values from the df
null_df=new_df.dropna()
```

```python
# it will delete selected column null values from the df
null_df=new_df.dropna(subset=["Is_Active"])
null_df.select("Student_ID","Age","Is_Active").display()
```

```python
# this will fill the null valuse with the specific value
null_df=new_df.fillna(999, subset=["Is_Active"])
null_df.select("Student_ID","Age","Is_Active").display()
```
